```python
#https://codeforces.com/group/MWSDmqGsZm/contest/219432/problem/Q



t = int(input())

for i in range(t):
    number = int(input())
    if number == 0:
        print(0)
        continue
    while number > 0:
        digit = number % 10
        print(digit, end=" ")
        number //= 10    #convert number to integer and remove the last digit
    print()
```
