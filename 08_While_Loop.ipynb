```python
count = int (input("Enter the Start Number = "))

while count < 10:
    if count == 0:
        print(count," is Zero Number!")
    elif count % 2 == 0:
        print(count, " is Even Number!")
    else:    
        print(count, " is Odd Number!")
    count = count + 1    
```
