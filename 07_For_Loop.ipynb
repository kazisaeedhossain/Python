```python
# Example 1: range with only stop argument
for i in range(5):
    print(i)  # Output: 0, 1, 2, 3, 4
```

```python
# Example 2: range with start and stop arguments
for i in range(2, 5):
    print(i)   # Output: 2, 3, 4
```

```python
# Example 3: range with start, stop, and step arguments
for i in range(1, 10, 2):
    print(i)  # Output: 1, 3, 5, 7, 9
```

```python
# Example 4: Print numbers from 1 to 5 using a for loop
for num in range(1, 6):
    print(num)
```

```python
x = int(input("Enter the Start Number = "))
y = int(input("Enter the End Number = "))

   
for num in range(x,y):
    if num == 0:
            print(num, " is Zero ")
    elif num % 2 == 0:
           print(num, " is even")
    else:
            print(num, " is odd")
            
```
