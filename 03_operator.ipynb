```python
first_number = 10
second_number = 5

sum = first_number + second_number     # Summation
sub = first_number - second_number     # Subtraction
mult = first_number * second_number    # Multiplication
div = first_number / second_number     # Division
mod = first_number % second_number     # Modulation

print("Summation = " + str(sum))
print("Subtraction = " + str(sub))
print("Multiplication = " + str(mult))
print("Division = " + str(int(div)))
print("Modulation = " + str(mod))
```
