```python
string= """The Name Of Our Country IS Bangladesh.
I Name of our City Is Narayanganj.
My Name is Kazi Saeed Hossain"""

print(string)
print("\n")
print ("Length of String = ",len(string))
print("\n")

process_string = string.lower()
print(process_string)
print("\n")

#word ase kina -> Bool
print("Bangladesh" in string )
print("bangladesh" in process_string)
print("\n")

#shurur dike kon index theke  substring suru  hoise
index = process_string.find("name")
print(index)


#Sheser kon index theke subtring suru hoise
reverse_index = process_string.rfind("name")
print(reverse_index)


#Subtring koy bar ache
count = process_string.count("name")
print(count)
print("\n")

replace_string = string.replace("Kazi Saeed Hossain","Kazi Nayeem Islam")
print(replace_string)
```
